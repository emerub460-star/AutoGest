## Problemas de calidad
Se listan los 3+ problemas exigidos por la guía:

````python
# 1. Columna 100% nula
pct = df['Customer Feedback'].isna().mean()*100
print(f'Customer Feedback: {pct:.1f}% nula')

# 2. Columna corrupta
n = df['Service Status'].value_counts().get('s', 0)
print(f'Service Status: {n:,} filas corruptas')

# 3. Nulos estructurales MAR (grúa)
print(df.isna().sum()[df.isna().sum()>0].to_string())

# 4. Valor piso sospechoso
print('Total Cost == $20:', int((df['Total Cost']==20).sum()), 'filas')
````
